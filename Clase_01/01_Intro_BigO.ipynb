## 2. Abstracción: Procedural y de Datos (ADT)

La base fundamental para diseñar buenas estructuras de datos es la **Abstracción**. En informática, la abstracción consiste en ocultar los detalles complejos de implementación al usuario final, mostrándole únicamente la funcionalidad esencial.

Existen dos tipos principales de abstracción:
1.  **Procedural Abstraction (Abstracción Procedimental):** Es ocultar el algoritmo o los pasos específicos. Cuando usted usa la función `Math.sqrt(25)` en Java, a usted no le importa si usa el método babilónico o la serie de Taylor por debajo; solo le importa que devuelva `5.0`.
2.  **Data Abstraction (Abstracción de Datos):** Es ocultar cómo se almacena físicamente la información en la memoria (por ejemplo, si usamos un arreglo o un sistema de nodos enlazados).

### El Tipo de Dato Abstracto (Abstract Data Type - ADT)
Cuando combinamos la abstracción de datos y la procedimental, creamos un **ADT**. Un ADT es un modelo lógico que deﬁne qué datos contiene una estructura y qué operaciones se pueden hacer sobre ellos, pero **no** dice cómo están programados.

**Analogía del Automóvil:**
Para conducir un automóvil, usted no necesita saber si el motor es de cuatro cilindros o rotativo (abstracción de datos), ni si la dirección es mecánica o eléctrica (abstracción procedimental). Usted solo sabe que si presiona el acelerador, el auto avanza, y si gira el volante a la izquierda, el auto gira a la izquierda. El automóvil es un ADT.

## 3. Interfaces en Java: Separando el "Qué" del "Cómo"

En Java, la mejor forma de representar un ADT es mediante una **Interfaz** (`interface`). Una interfaz lista los encabezados de los métodos (el *qué* debe hacer) sin incluir el código interno (el *cómo* lo hace).

Consideremos un programa sencillo para almacenar puntajes y calcular el promedio. Podemos deﬁnir una interfaz llamada `ScoresAvgADT`.

````python
%%writefile ScoresAvgADT.java
// Esta interfaz define el ADT. Es el contrato que cualquier clase debe cumplir.
public interface ScoresAvgADT {
    // Solo definimos QUÉ debe poder hacer, no CÓMO lo hace.
    void enterScores();
    void computeAverage();
}
````

````python
%%writefile Scores1D.java
import java.util.Scanner;

// Esta clase es la IMPLEMENTACIÓN del ADT.
// Aquí el programador decidió usar un arreglo (Data Abstraction)
// y un ciclo for (Procedural Abstraction).
public class Scores1D implements ScoresAvgADT {
    private double[] scores;
    private final int NUM_EXAMS = 3;

    public Scores1D() {
        scores = new double[NUM_EXAMS];
    }

    public void enterScores() {
        scores[0] = 4.5;
        scores[1] = 3.8;
        scores[2] = 5.0;
        System.out.println("Puntajes ingresados en el arreglo.");
    }

    public void computeAverage() {
        double sum = 0;
        for(int i = 0; i < NUM_EXAMS; i++) {
            sum += scores[i];
        }
        System.out.printf("El promedio es: %.2f\n", sum / NUM_EXAMS);
    }
}
````

````python
%%writefile MainApp.java
public class MainApp {
    public static void main(String[] args) {
        // El usuario del ADT interactúa con la interfaz.
        ScoresAvgADT misPuntajes = new Scores1D();

        misPuntajes.enterScores();
        misPuntajes.computeAverage();
    }
}
````

## 4. Memoria Estática vs. Dinámica

En la clase anterior (`Scores1D`), tomamos una decisión de diseño: usamos un arreglo (`double[]`). Los arreglos son estructuras **Estáticas**.

*   **Estático (Static):** Su tamaño es ﬁjo y no puede cambiar durante la ejecución del programa. Son muy rápidos, pero consumen memoria por adelantado y no son ﬂexibles.
*   **Dinámico (Dynamic):** Usan referencias y objetos (como las listas enlazadas que veremos más adelante). Pueden crecer o encogerse durante la ejecución del programa, pero consumen más memoria RAM por cada nodo y pueden ser ligeramente más lentas de procesar.

## 5. Análisis Algorítmico Elemental (Contando Operaciones)

El hecho de que un programa compile no signiﬁca que sea un *buen* programa. Si diseñamos una estructura de datos para procesar 10 elementos, cualquier algoritmo sirve. Pero en el mundo real, podríamos procesar **100 millones de registros**.

El **Análisis de Algoritmos** busca determinar el rendimiento de un código sin importar qué tan rápida sea la computadora (CPU) que lo ejecuta. Para ello, contamos las "unidades de tiempo" u operaciones fundamentales que realiza un bloque de código en función del tamaño de la entrada, la cual llamaremos **$n$**.

### Ejemplo de Conteo:
Supongamos el siguiente bucle:
```java
i = 1;                  // 1 unidad de tiempo
while(i <= n) {         // se evalúa (n+1) veces
    total = total + x;  // se ejecuta n veces
    i = i + 1;          // se ejecuta n veces
}
```
El total de unidades de tiempo para este código es: $1 + (n+1) + n + n = 3n + 2$.

¿Qué pasa si añadimos una variable más dentro del bucle? El tiempo total podría subir a $4n + 2$. La diferencia entre $3n+2$ y $4n+2$ es despreciable cuando $n$ vale 1 millón. Por esta razón, el análisis algorítmico se enfoca en el **comportamiento Asintótico**: qué sucede cuando $n$ tiende a inﬁnito.

## 6. Notación Big O y Simplificación Matemática

La notación **Big O** ($O$) es el estándar matemático que clasiﬁca a los algoritmos según su tasa de crecimiento en el *peor de los casos*.

### Reglas de Simplificación (Reglas de Oro):
Al convertir una función de tiempo exacto a Notación Big O, aplicamos dos reglas:
1.  **Ignorar Coeﬁcientes y Constantes:** Si calculamos $O(3n + 2)$, descartamos el multiplicador 3 y la constante 2. Nos queda simplemente **$O(n)$**.
2.  **Mantener el Término Dominante (El más lento):** Si tenemos un algoritmo complejo que tarda $n^2 + n + 500$ operaciones, cuando $n$ es enorme (ej. 1,000,000), el $n^2$ domina abrumadoramente el tiempo de cálculo. Por ende, $O(n^2 + n + 500)$ se simpliﬁca a **$O(n^2)$**.

### Categorías de Complejidad Comunes:
*   **$O(1)$ - Constante:** El tiempo de ejecución nunca cambia. (Ej. Acceder a un arreglo por su índice `datos[5]`).
*   **$O(\log n)$ - Logarítmica:** Extremadamente rápido. Típico de algoritmos que dividen el problema a la mitad en cada paso.
*   **$O(n)$ - Lineal:** El tiempo crece proporcionalmente a $n$. Típico de un bucle `for` simple.
*   **$O(n \log n)$ - Lineal-Logarítmica:** Típico de los mejores algoritmos de ordenamiento moderno.
*   **$O(n^2)$ - Cuadrática:** Lento. Típico de bucles anidados. Peligroso con datos muy grandes.

## 7. Demostración en Vivo: Midimiendo Tiempos con System.nanoTime()

Vamos a experimentar esto en la realidad. Escribiremos un programa en Java que procesará un arreglo de 20,000 elementos de dos formas:
1.  Con un algoritmo lineal **$O(n)$** (imprimir solo los 5 primeros elementos, pero recorriendo todos por debajo para sumar algo inútil).
2.  Con un algoritmo cuadrático **$O(n^2)$** (dos bucles anidados procesando los mismos 20,000 elementos).

Usaremos `System.nanoTime()` para medir los microsegundos reales de la CPU de Google Colab.

````python
%%writefile ComparadorDeComplejidad.java
public class ComparadorDeComplejidad {

    public static void main(String[] args) {
        int n = 20000; // Tamaño de entrada respetable
        int[] datos = new int[n];

        // Llenamos el arreglo con unos
        for(int i=0; i<n; i++) {
            datos[i] = 1;
        }

        System.out.println("Iniciando prueba con N = " + n + "\n");

        // PRUEBA 1: O(n) - Lineal
        long inicioLineal = System.nanoTime();

        long sumaLineal = 0;
        for(int i=0; i<n; i++) {
            sumaLineal += datos[i]; // Un bucle simple
        }

        long finLineal = System.nanoTime();
        double tiempoLinealMs = (finLineal - inicioLineal) / 1_000_000.0;
        System.out.printf("Tiempo O(n): %.4f milisegundos\n", tiempoLinealMs);

        // PRUEBA 2: O(n^2) - Cuadrático
        long inicioCuadratico = System.nanoTime();

        long sumaCuadratica = 0;
        for(int i=0; i<n; i++) {
            for(int j=0; j<n; j++) {
                sumaCuadratica += (datos[i] * datos[j]); // Bucle anidado
            }
        }

        long finCuadratico = System.nanoTime();
        double tiempoCuadraticoMs = (finCuadratico - inicioCuadratico) / 1_000_000.0;
        System.out.printf("Tiempo O(n^2): %.4f milisegundos\n", tiempoCuadraticoMs);

        System.out.println("\nCONCLUSIÓN:");
        System.out.printf("El algoritmo O(n^2) tardó %.0f veces más que el O(n).\n", (tiempoCuadraticoMs / tiempoLinealMs));
    }
}
````

````python
!javac ComparadorDeComplejidad.java
!java ComparadorDeComplejidad
````

**¡Atención a la Conclusión!**
Con un tamaño de apenas 20,000 elementos, el algoritmo cuadrático demoró exponencialmente más. Si usted cambia el tamaño a 1,000,000 en un proyecto real, un algoritmo cuadrático podría colapsar el servidor, mientras que el lineal tardaría fracciones de segundo. ¡Por eso es vital dominar las Estructuras de Datos y el Big O!

## 8. Optimizaciones en la Vida Real

Uno de los recursos más interactivos para entender visualmente el impacto de estas complejidades es el artículo **[A visual introduction to Big O notation](https://samwho.dev/big-o/)** por Sam Rose. De él rescatamos los siguientes problemas clásicos en la ingeniería de software:

### El costo oculto en los bucles
¿Qué sucede si dentro de un bucle `for` inocente llamamos a una función que también recorre toda la lista (como `indexOf` o `contains` en listas no optimizadas)?
Si usamos un método $O(n)$ dentro de un bucle que ya es $O(n)$, el costo total de la función se dispara mágicamente a **$O(n^2)$**. En la vida real, muchos desarrolladores cometen el error de escribir cosas como:

```java
for (String item : lista) {
    int indice = lista.indexOf(item); // ¡PELIGRO! indexOf es O(n)
    System.out.println("Item " + indice + ": " + item);
}
```
Para corregir esto, iterar directamente usando el índice `i` reduce la complejidad a $O(n)$ porque el acceso a un índice específico en un arreglo es $O(1)$.

### Suma Constante (El truco de Gauss)
Si le piden sumar todos los números del 1 al $n$, la intuición natural es usar un bucle iterativo que tarde $O(n)$. Sin embargo, las matemáticas nos dan la fórmula `(n * (n + 1)) / 2`. Usar una fórmula matemática pura cambia la complejidad de $O(n)$ a **$O(1)$**, haciendo que el resultado sea instantáneo incluso si $n$ vale 100 billones.

### Conclusión
La notación Big O siempre evalúa el **peor caso**. Un algoritmo $O(n)$ podría por suerte encontrar la respuesta en su primer paso $O(1)$, pero a los ingenieros de software se les paga por garantizar el sistema en las peores condiciones de tráfico.

## 9. Ejercicios de Análisis de Complejidad

Analice los siguientes fragmentos de código e intente deducir su complejidad **Big O**.

**Fragmento A:**
```java
for(i=1; i<=n; i++) {
    for(j=1; j<=n; j++) {
        // algunas instrucciones
    }
}
for(i=1; i<=n; i++) {
    // algunas instrucciones
}
```
*(Análisis: Los primeros ciclos anidados tardan $n \times n = n^2$. El último ciclo tarda $n$. El total es $O(n^2 + n)$. Siguiendo las reglas de oro y manteniendo el dominante, esto se simplifica a **$O(n^2)$**).*

**Fragmento B:**
```java
i = n;
while(i > 1) {
    // algunas instrucciones
    i = i / 2;
}
```
*(Análisis: Si $n$ vale 16, $i$ toma los valores 16, 8, 4, y 2. Iteró 4 veces. Matemáticamente $2^4 = 16$. Por ende, la complejidad es logarítmica, **$O(\log n)$**).*
